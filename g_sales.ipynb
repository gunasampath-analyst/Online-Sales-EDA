# 📄 Project Summary

This project performs Exploratory Data Analysis (EDA) on an Online Sales dataset to uncover business insights related to sales performance, profitability, customer behavior, discounts, shipping costs, and regional performance.

# 🎯 Problem Statement

Businesses collect large amounts of sales data every day, but raw data alone does not provide meaningful insights. This project analyzes online sales transactions to identify sales trends, profitable products, customer behavior, and operational opportunities that can improve business performance.

---

The objectives of this analysis are to:

- Analyze overall sales and profit performance.
- Identify profitable and loss-making categories and products.
- Evaluate the impact of discounts on profitability.
- Analyze customer purchasing behavior.
- Compare market and regional performance.
- Study shipping costs.
- Generate actionable business insights.

## 📚 Import Required Libraries

```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## 📥 Load the Dataset


```python
sale = pd.read_csv("Global Superstore.csv")
sale.head()
```

## 📏 Dataset Dimensions

```python
rows , col = sale.shape
print(f"There are '{rows}' rows and '{col}' columns in the dataset." )
```

## ℹ️ Dataset Information

```python
sale.info()
```

## 🔍 Missing Value Analysis

```python
sale.isnull().sum()
```

## 🧹 Duplicate Records

```python
Duplicates = sale.duplicated().sum()
print(f"There are '{Duplicates}' in this dataset.")
```

## 📊 Statistical Summary

```python
sale.describe()
```

## 🌍 Top 10 Countries by Sales

```python
country_sales = sale.groupby("Country")["Sales"].sum().sort_values(ascending=False).head(10)
print(country_sales)
```

```python
plt.figure(figsize=(8,6))
sns.barplot(x=country_sales.index,
            y=country_sales.values,
            palette="rocket" ,
            width=0.6
)
plt.title("Top 10 Countries by Sales",fontsize=15,fontweight="bold")

plt.xlabel("Sales ($)")
plt.ylabel("Country")

plt.xticks(rotation=90)
plt.ylim(0,country_sales.max()*1.1)

plt.grid(axis="y",linestyle="--",alpha=.4 , color="black")

plt.tight_layout()
plt.show()
```

## 📦 Sales by Category

```python
category_sales = sale.groupby("Category")["Sales"].sum().sort_values(ascending=False)
print(category_sales)
```

```python
colors = sns.color_palette("Set2")

plt.pie(category_sales,
        labels=category_sales.index,
        autopct="%1.2f%%",
        startangle=90,
        colors=colors,
        wedgeprops={"width":0.45},
        pctdistance=0.77)

plt.title("Sales Contribution by Category",fontsize=15,fontweight="bold")

plt.tight_layout()
plt.show()
```

## 🗂️ Sales by Sub-Category

```python
subcategory_sales = sale.groupby("Sub-Category")["Sales"].sum().sort_values(ascending=False)
print(subcategory_sales)

```

```python
plt.figure(figsize=(8,6))
sns.barplot(x=subcategory_sales.values,
            y=subcategory_sales.index,
            palette='crest'
)

plt.title("Top 10 Sub-Categories by Sales",fontsize=15,fontweight="bold")

plt.xlabel("Sales ($)")
plt.ylabel("Sub-Category")

plt.xlim(0,subcategory_sales.max()*1.1)
plt.grid(axis='x',linestyle='--',alpha=.4)
plt.tight_layout()
plt.show()
```

## 🛍️ Top 10 Products by Sales

```python
product_sales = sale.groupby("Product Name")["Sales"].sum().sort_values(ascending=False).head(10)
print(product_sales)
```

```python
sns.barplot(x=product_sales.values,
            y=product_sales.index,
            palette="Reds",
            edgecolor="black",
            linewidth=0.5 
)
plt.title("Top 10 Products" , fontsize=14,fontweight="bold")
plt.xlabel("Total Profit ($)")
plt.show()
```

## 🌐 Sales by Market

```python
market_sales = sale.groupby("Market")["Sales"].sum().sort_values(ascending=False).head(5)
print(market_sales)
```

```python
plt.figure(figsize=(6, 6))

plt.pie(
    market_sales,
    labels=market_sales.index,
    autopct="%1.1f%%",
    startangle=90,
    colors=sns.color_palette("Set2"),
    wedgeprops={"edgecolor": "black" , "linewidth": 0.5 ,"linestyle" : "--"}
)

plt.title("Sales Contribution by Market", fontweight="bold")
plt.tight_layout()
plt.show()
```

## 💰 Profit by Market

```python
market_profit = sale.groupby("Market")["Sales"].sum().sort_index()
print(market_profit)
```

```python
sns.barplot(y=market_profit.index,
         x=market_profit.values,
         palette="Reds",
         edgecolor="black",
         linewidth=0.5
         )
plt.title("Profit Generated by Market" , fontsize=14,fontweight="bold")
plt.xlabel("Profit ($)")
plt.ylabel("Market")

plt.tight_layout()
plt.show()
```

## 🚚 Order Distribution by Ship Mode

```python
ship_mode_orders = (sale["Ship Mode"].value_counts().sort_values(ascending=False))
print(ship_mode_orders)
```

```python
sns.histplot(data=sale,x="Ship Mode",stat="count",shrink=0.8,color="#30DED8",edgecolor="black")

plt.title("Order Count by Ship Mode", fontsize=14, fontweight="bold", pad=15)
plt.xlabel("Ship Mode", fontsize=11, fontweight="bold")
plt.ylabel("Number of Orders", fontsize=11, fontweight="bold")
plt.grid(alpha=0.5)
plt.tight_layout()
plt.show()
```

## 📈 Yearly Profit by Category

```python
category_profit = sale.groupby(["Year","Category"])["Profit"].sum().reset_index()
print(category_profit)
```

```python
sns.lineplot(data=category_profit,
             x="Year",
             y="Profit",
             hue="Category",
             marker="o",
             linestyle="--",
             linewidth=1.5)
plt.title("Profit Trend by Category" , fontsize=14 , fontweight="bold")
plt.xlabel("Year")
plt.ylabel("Total Profit ($)")
plt.grid(alpha=0.5)
plt.tight_layout()
plt.show()
```

## 📉 Discount Distribution

```python
plt.figure(figsize=(8,5))

sns.histplot(
    data=sale,
    x="Discount",
    bins=20,
    kde=True,
    color="steelblue"
)

plt.title("Discount Distribution", fontsize=15, fontweight="bold")
plt.xlabel("Discount")
plt.ylabel("Number of Orders")

plt.grid(axis="y", alpha=0.3)
plt.show()
```

## 🚢 Shipping Cost by Market

```python
total_shipping_market = sale.groupby("Market")["Shipping Cost"].sum().sort_values(ascending=False)
print(total_shipping_market)
```

```python
sns.barplot(x=total_shipping_market.index,
            y=total_shipping_market.values,
            edgecolor="black",
            linewidth=0.5,
            palette="Greens",
            
)
plt.title("Total Shipping Cost by Market",fontsize=14,fontweight="bold")
plt.xlabel("Market",fontweight="demibold")
plt.ylabel("Total Shipping Cost ($)" , fontweight="demibold")
plt.tight_layout()
plt.show()
```

## 📉 Top Loss-Making Products


```python
loss_products = sale.groupby("Product Name")["Profit"].sum().sort_values(ascending=False)

loss = loss_products[loss_products<0].head(10)

print(loss)
```

```python
sns.barplot(
    x=loss.values,
    y=loss.index,
    palette="Blues",
    edgecolor="black",
    linewidth=0.5
)
plt.title("Top 10 Loss-Making Products" , fontsize=14 , fontweight="bold")
plt.xlabel("Total Profit ($)")
plt.show()
```

## 📊 Sales vs Profit Analysis

```python
plt.figure(figsize=(8,6))
sns.scatterplot(data=sale,
                x="Sales",
                y="Profit",
                hue="Category"             
)
plt.title("Sales vs Profit" , fontsize=14 , fontweight = "bold")
plt.xlabel("Sales ($)")
plt.ylabel("Profit ($)")
plt.axhline(color="red",linestyle="--" , linewidth=0.9)
plt.legend(loc="best")
plt.grid(alpha=0.5)
plt.show()
```

## 🔗 Correlation Analysis

```python
col = ["Sales","Profit","Quantity","Shipping Cost"]
corr = sale[col].corr()
print(corr)
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    corr,
    annot=True,
    cmap="mako",
    fmt=".2f",
    linewidths=0.5,
    square=True
)

plt.title("Correlation Heatmap", fontsize=15, fontweight="bold")

plt.show()
```

## 📈 Profit Margin Analysis

```python
sale["Profit Margin"] = sale["Profit"] / sale["Sales"] * 100
print(sale["Profit Margin"])
```

## 📊 Average Profit Margin by Category

```python
profit_margin_category  = sale.groupby("Category")["Profit Margin"].mean().sort_values(ascending=False)
print(profit_margin_category)
```

```python
sns.barplot(x=profit_margin_category.index,
            y=profit_margin_category.values,
            palette="magma",
            edgecolor="black",
            linewidth=0.5
)
plt.title("Average Profit Margin by Category", fontsize=14 , fontweight="bold")
plt.xlabel("Category")
plt.ylabel("Profit Margin (%")
plt.grid(axis="y",alpha=0.3 , linestyle="-." , linewidth=0.7,color="black")
plt.ylim(0,profit_margin_category.max()*1.1)
plt.tight_layout()
plt.show()
```

## 🌍 Average Profit Margin by Market

```python
profit_margin_market = sale.groupby("Market")["Profit Margin"].mean().sort_values(ascending=False)
print(profit_margin_market)
```

```python
sns.barplot(x=profit_margin_market.values,
            y=profit_margin_market.index,
            palette="viridis",
            edgecolor="black",
            linewidth=0.5)
plt.title("Average Profit Margin by Market",fontsize=14,fontweight="bold")
plt.xlabel("Profit Margin (%)")
plt.ylabel("Market")
plt.grid(axis="x",alpha=0.3,linestyle="--")
plt.tight_layout()
plt.show()
```

## 🌍 Region-wise Sales and Profit Analysis

```python
region_sales = sale.groupby("Region")["Sales"].sum().reset_index()
print(region_sales)
```

```python
region_profit = sale.groupby("Region")["Profit"].sum().reset_index()
print(region_profit)
```

```python
fig, (ax1, ax2) = plt.subplots(2,1,figsize=(15,8))
sns.lineplot(data=region_sales,
             x="Region",
             y="Sales",
             marker="o",
             linewidth=2,
             color="#1f77b4",
             ax=ax1)
ax1.set_title("Region-wise Sales",fontsize=14 , fontweight="bold")
ax1.set_xlabel("")
ax1.set_ylabel("Sales")
ax1.grid(alpha=0.3)

sns.lineplot(data=region_profit,
            x="Region",
            y="Profit",
            marker="o",
            linewidth=2,
            color="#2ca02c",
            ax=ax2)
ax2.set_title("Region-wise Profit", fontsize=14,fontweight="bold")
ax2.set_xlabel("Region")
ax2.set_ylabel("Profit")
ax2.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```
